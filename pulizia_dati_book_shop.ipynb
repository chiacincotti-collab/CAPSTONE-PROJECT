```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("sheelakasbekar/books-data-set")

print("Path to dataset files:", path)
```

```python
import os
os.listdir(path)
```

```python
import pandas as pd
file1 = os.path.join(path, 'Books.csv')
books = pd.read_csv(file1)
books.head()
```

```python
books.info()
```

```python
books.loc[books.duplicated()]
```

```python
books.loc[books["Book_ID"].isna()]
```

```python
file2 = os.path.join(path, 'Customers.csv')
customers = pd.read_csv(file2)
customers.head()
```

```python
customers.info()
```

```python
customers.loc[customers.duplicated()]
```

```python
customers.loc[customers["Customer_ID"].isna()]
```

```python
file3 = os.path.join(path, 'Orders.csv')
orders = pd.read_csv(file3)
orders.head()
```

```python
orders.info()
```

```python
orders.loc[orders.duplicated()]
```

```python
orders.loc[orders["Order_ID"].isna()]
```

```python
orders["Order_Date"] = pd.to_datetime(orders["Order_Date"])
orders.info()
```

```python
books.to_csv("books_pulita.csv", index=False)
customers.to_csv("customers_pulita.csv", index=False)
orders.to_csv("orders_pulita.csv", index=False)
```
